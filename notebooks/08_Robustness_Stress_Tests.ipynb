## 1. Drive authentication and operational bounds


```python
PROJECT_DIRECTORY='/content/drive/MyDrive/DiffusionPolicyEvents'
AUTO_DISCONNECT_ON_COMPLETE=True
EXPORT_FULL_BUNDLE=False  # Packaging only; original files stay on Drive.
IO_NO_PROGRESS_SECONDS=120
SILENT_STAGE_NO_PROGRESS_SECONDS=600

from google.colab import drive
drive.mount('/content/drive')
import base64, hashlib, importlib.util, json, os, sys, tempfile, zlib
from pathlib import Path
RECOVERY_LOCAL=Path(tempfile.mkdtemp(prefix='nb08-io-recovery-',dir='/content'))
print('Local recovery logs:',RECOVERY_LOCAL)
print('Keep this runtime alone; stop/disconnect any previous final session.')
```

## 2. Unchanged scientific payload and separate I/O supervisor


```python
RECOVERY_PAYLOAD=json.loads(zlib.decompress(base64.b64decode('')))
OPERATION_SHA256='47ec2d84b754d2feab6d40b08d5a743148b2ad3ccdd00bb0b4a47616474d871e'
FROZEN_BUILD_SHA256='be9e982886f40ca61e6348b894c3961641a4298806125794a4ff784857177b25'
STUDY_ID='d200_tail_final_normal_dark_v1_fed681d7b085'
assert hashlib.sha256(json.dumps(RECOVERY_PAYLOAD,sort_keys=True,separators=(',',':')).encode()).hexdigest()==OPERATION_SHA256
for key,name in [('frozen_configuration','frozen_configuration.py'),('frozen_payload','frozen_payload.py'),('supervisor','io_supervisor.py')]:
    text=RECOVERY_PAYLOAD[key]
    compile(text,name,'exec')
    (RECOVERY_LOCAL/name).write_text(text)
EXPECTED_OUTPUT=PROJECT_DIRECTORY+'/results/final_evaluation/'+STUDY_ID
REQUEST={'project_root':PROJECT_DIRECTORY,'expected_output':EXPECTED_OUTPUT,
    'operation_sha256':OPERATION_SHA256,'frozen_build_sha256':FROZEN_BUILD_SHA256,
    'io_timeout_seconds':IO_NO_PROGRESS_SECONDS,
    'silent_stage_timeout_seconds':SILENT_STAGE_NO_PROGRESS_SECONDS,
    'export_full_bundle':EXPORT_FULL_BUNDLE,
    'known_completed_keys':RECOVERY_PAYLOAD['known_completed_keys']}
(RECOVERY_LOCAL/'request.json').write_text(json.dumps(REQUEST,indent=2))
spec=importlib.util.spec_from_file_location('nb08_operational_supervisor',RECOVERY_LOCAL/'io_supervisor.py')
RECOVERY=importlib.util.module_from_spec(spec)
spec.loader.exec_module(RECOVERY)
print('Frozen scientific build:',FROZEN_BUILD_SHA256)
print('Separate operational wrapper:',OPERATION_SHA256)
print('Same study and output:',EXPECTED_OUTPUT)
```
